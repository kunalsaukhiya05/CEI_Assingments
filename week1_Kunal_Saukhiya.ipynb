### 1.1 Data Types & Control Flow

Write a function `classify_number(n)` that:
- Returns `"negative"` if `n < 0`
- Returns `"zero"` if `n == 0`
- Returns `"small positive"` if `0 < n <= 10`
- Returns `"large positive"` otherwise

````python
def classify_number(n):
    if n < 0:
        return "negative"
    elif n == 0:
        return "zero"
    elif n <= 10:
        return "small positive"
    else:
        return "large positive"

# Tests
assert classify_number(-5) == "negative"
assert classify_number(0)  == "zero"
assert classify_number(7)  == "small positive"
assert classify_number(42) == "large positive"
print("1.1 passed ✓")
````

### 1.2 Data Structures

Given the list `words = ['apple', 'banana', 'cherry', 'apple', 'date', 'banana', 'apple']`:

1. Build a **dict** `word_count` mapping each word to its frequency using only a loop (no `Counter`).
2. Using a **set**, find all unique words and store in `unique_words`.
3. Using a **list comprehension**, build `long_words` containing words with more than 5 characters.

````python
words = ['apple', 'banana', 'cherry', 'apple', 'date', 'banana', 'apple']

# 1. word_count dict
word_count = {}
for word in words:
    word_count[word] = word_count.get(word, 0) + 1

# 2. unique_words set
unique_words = set(words)

# 3. long_words list comprehension
long_words = [w for w in words if len(w) > 5]

# Tests
assert word_count == {'apple': 3, 'banana': 2, 'cherry': 1, 'date': 1}
assert unique_words == {'apple', 'banana', 'cherry', 'date'}
assert set(long_words) == {'banana', 'cherry'}
print("1.2 passed ✓")
````

### 1.3 Exceptions

Write `safe_divide(a, b)` that:
- Returns `a / b` normally
- Returns `None` if `b == 0` (catch `ZeroDivisionError`)
- Raises `TypeError` with message `"Inputs must be numeric"` if either input is not a number

````python
def safe_divide(a, b):
    if not (isinstance(a, (int, float)) and isinstance(b, (int, float))):
        raise TypeError("Inputs must be numeric")
    try:
        return a / b
    except ZeroDivisionError:
        return None

# Tests
assert safe_divide(10, 2) == 5.0
assert safe_divide(5, 0) is None
try:
    safe_divide('x', 2)
    assert False, "Should have raised TypeError"
except TypeError as e:
    assert str(e) == "Inputs must be numeric"
print("1.3 passed ✓")
````

````python
def apply_twice(f, x):
    return f(f(x))

triple = lambda x: x * 3

result = apply_twice(triple, 4)
assert result == 36, f"Expected 36, got {result}"
print("1.4 passed ✓")
````

## Part 4 — Linear Algebra

### 4.1 Matrix Inverse & Determinant

Given:
```
M = [[4, 7], [2, 6]]
```

1. Compute **determinant** → `det_M`
2. Compute **inverse** → `inv_M`
3. Verify: `M @ inv_M ≈ I` (identity matrix)

````python
M = np.array([[4, 7], [2, 6]])

det_M = np.linalg.det(M)
inv_M = np.linalg.inv(M)
I = M @ inv_M

assert abs(det_M - 10.0) < 0.0001
assert np.allclose(I, np.eye(2))
print("4.1 passed ✓")
print(f"Determinant: {det_M:.3f}")
print(f"Inverse:\n{inv_M}")
print(f"M @ M^-1 ≈ I:\n{I}")
````

### 4.2 Eigenvalues & Eigenvectors

1. Compute eigenvalues and eigenvectors of `M`
2. Store eigenvalues in `eigenvals` and eigenvectors in `eigenvecs`

````python
eigenvals, eigenvecs = np.linalg.eig(M)

assert len(eigenvals) == 2
assert eigenvecs.shape == (2, 2)
print("4.2 passed ✓")
print(f"Eigenvalues: {eigenvals}")
print(f"Eigenvectors:\n{eigenvecs}")
````

### 5.2 Normal Distribution

1. Generate 10,000 samples from a normal distribution with μ=100, σ=15
2. Compute the **z-score** of value 130 → `z_score`
3. Find the **percentile** of z-score using `scipy.stats.norm.cdf()` → `percentile`

````python
samples = np.random.normal(100, 15, 10000)
value = 130
z_score = (value - 100) / 15
percentile = stats.norm.cdf(z_score)

assert -0.1 < samples.mean() - 100 < 0.1  # Should be close to 100
assert abs(z_score - 2.0) < 0.0001
assert 0.97 < percentile < 0.99  # ~97.7%
print("5.2 passed ✓")
print(f"Sample mean: {samples.mean():.2f}")
print(f"Sample std: {samples.std():.2f}")
print(f"Z-score for 130: {z_score:.3f}")
print(f"Percentile: {percentile*100:.2f}%")
````

### 5.3 Probability Theory

**Scenario:** A bag contains 5 red balls, 3 blue balls, 2 green balls.

1. Probability of drawing a red ball → `p_red`
2. Probability of drawing a blue ball → `p_blue`
3. Probability of drawing red OR blue → `p_red_or_blue`

````python
red, blue, green = 5, 3, 2
total = red + blue + green

p_red = red / total
p_blue = blue / total
p_red_or_blue = (red + blue) / total

assert abs(p_red - 5/10) < 0.0001
assert abs(p_blue - 3/10) < 0.0001
assert abs(p_red_or_blue - 8/10) < 0.0001
print("5.3 passed ✓")
print(f"P(red): {p_red:.3f}")
print(f"P(blue): {p_blue:.3f}")
print(f"P(red OR blue): {p_red_or_blue:.3f}")
````
