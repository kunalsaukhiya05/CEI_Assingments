### 1.1 Data Types and Control Flow

Write a function `classify_number(n)` that:
- Returns 'negative' if n < 0
- Returns 'zero' if n == 0
- Returns 'small positive' if 0 < n <= 10
- Returns 'large positive' otherwise

```python
def classify_number(n):
    if n < 0:
        return 'negative'
    elif n == 0:
        return 'zero'
    elif n <= 10:
        return 'small positive'
    else:
        return 'large positive'

assert classify_number(-5) == 'negative'
assert classify_number(0) == 'zero'
assert classify_number(7) == 'small positive'
assert classify_number(42) == 'large positive'
print('1.1 passed')
```

### 1.2 Data Structures

Given: words = ['apple', 'banana', 'cherry', 'apple', 'date', 'banana', 'apple']

1. Build a dict mapping each word to frequency
2. Find unique words using a set
3. Build list of words with more than 5 characters

```python
words = ['apple', 'banana', 'cherry', 'apple', 'date', 'banana', 'apple']

word_count = {}
for word in words:
    word_count[word] = word_count.get(word, 0) + 1

unique_words = set(words)
long_words = [w for w in words if len(w) > 5]

assert word_count == {'apple': 3, 'banana': 2, 'cherry': 1, 'date': 1}
assert unique_words == {'apple', 'banana', 'cherry', 'date'}
assert set(long_words) == {'banana', 'cherry'}
print('1.2 passed')
```

### 1.3 Exceptions

Write safe_divide(a, b) that:
- Returns a/b normally
- Returns None if b == 0
- Raises TypeError if inputs are not numeric

```python
def safe_divide(a, b):
    if not (isinstance(a, (int, float)) and isinstance(b, (int, float))):
        raise TypeError('Inputs must be numeric')
    try:
        return a / b
    except ZeroDivisionError:
        return None

assert safe_divide(10, 2) == 5.0
assert safe_divide(5, 0) is None
try:
    safe_divide('x', 2)
    assert False
except TypeError:
    pass
print('1.3 passed')
```

### 1.4 Functions and Lambdas

1. Write apply_twice(f, x) that applies f twice
2. Create triple lambda that multiplies by 3
3. apply_twice(triple, 4) should return 36

```python
def apply_twice(f, x):
    return f(f(x))

triple = lambda x: x * 3
result = apply_twice(triple, 4)
assert result == 36
print('1.4 passed')
```

## Part 4 - Linear Algebra

### 4.1 Matrix Inverse and Determinant

M = [[4, 7], [2, 6]]

1. Compute determinant
2. Compute inverse
3. Verify M @ M^-1 equals identity

```python
M = np.array([[4, 7], [2, 6]])

det_M = np.linalg.det(M)
inv_M = np.linalg.inv(M)
I = M @ inv_M

assert abs(det_M - 10.0) < 0.0001
assert np.allclose(I, np.eye(2))
print('4.1 passed')
```

### 4.2 Eigenvalues and Eigenvectors

Compute eigenvalues and eigenvectors of M

```python
eigenvals, eigenvecs = np.linalg.eig(M)

assert len(eigenvals) == 2
assert eigenvecs.shape == (2, 2)
print('4.2 passed')
```

### 5.2 Normal Distribution

1. Generate 10000 samples from N(100, 15)
2. Compute z-score for value 130
3. Find percentile using CDF

```python
samples = np.random.normal(100, 15, 10000)
value = 130
z_score = (value - 100) / 15
percentile = stats.norm.cdf(z_score)

assert -0.1 < samples.mean() - 100 < 0.1
assert abs(z_score - 2.0) < 0.0001
assert 0.97 < percentile < 0.99
print('5.2 passed')
```

### 5.3 Probability Theory

Bag contains 5 red, 3 blue, 2 green balls

1. Probability of red
2. Probability of blue
3. Probability of red OR blue

```python
red, blue, green = 5, 3, 2
total = red + blue + green

p_red = red / total
p_blue = blue / total
p_red_or_blue = (red + blue) / total

assert abs(p_red - 5/10) < 0.0001
assert abs(p_blue - 3/10) < 0.0001
assert abs(p_red_or_blue - 8/10) < 0.0001
print('5.3 passed')
```
